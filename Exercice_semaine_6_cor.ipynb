(a) Generate a simulated data set as follows:

```python
#Import numpy
import numpy as np

#Simulated data
rng = np.random.default_rng(1)
x = rng.normal(size=100)
y = x - 2 * x**2 + rng.normal(size=100)
```

In this dataset, n = 100 and p = 2. The model used to generate the data is:

$ Y = \beta_1X + \beta_2X^2 + \epsilon $

avec $\beta_1 = 1$ et $\beta_2 = -2$, et $\epsilon_i$ ~ $N(0,1)$.

(b) Create a scatterplot of X against Y . Comment on what you find.

```python
# Import matplotlib and seaborn packages
import matplotlib.pyplot as plt
import seaborn as sns

# Initialize figure
figsimu, axsimu = plt.subplots(1,1)
axsimu.set_title('Relationship between X and Y');

# Plot scatter and best fit line
sns.regplot(x=x, y=y, ax=axsimu, order=2, ci=None, scatter_kws={'color':'g', 's':10})
axsimu.legend(['Data','Least Square Fit']);
```

The regression line produced by a polynomial regression of degree 2 has a good fit because $y$ is a linear combination of $x$ and $x^2$. 

(c) Split your dataset in a training set and a test set using the random seed `450639`. Then compute the validation set errors that
result from fitting the following four models using least squares:

i. $Y = \beta_0 + \beta_1X + \epsilon$

ii. $Y = \beta_0 + \beta_1X + \beta_2X^2 + \epsilon$

iv. $Y = \beta_0 + \beta_1X + \beta_2X^2 + \beta_3X^3 + \beta_4X^4 + \epsilon$

```python
#Import required packages
import pandas as pd #To transform arrays in dataframes with pd.dataframe
from sklearn.model_selection import train_test_split #To perform sample splitting 
import statsmodels.api as sm #OLS regressions 
from ISLP.models import (ModelSpec as MS, #Data transformation
                         poly)


#Create dataframe and split it
df = pd.DataFrame({'x': x, 'y': y})

df_train, df_valid = train_test_split(df,
                                         test_size=50,
                                         random_state=450639)

#Program evaluation function
def evalMSE(terms,
            response,
            train,
            test):

   mm = MS(terms)
   X_train = mm.fit_transform(train)
   y_train = train[response]

   X_test = mm.transform(test)
   y_test = test[response]

   results = sm.OLS(y_train, X_train).fit()
   test_pred = results.predict(X_test)

   return np.mean((y_test - test_pred)**2)


#Obtain MSEs for each model
MSE = np.zeros(4)
for idx, degree in enumerate(range(1, 5)):
    MSE[idx] = evalMSE([poly('x', degree)],
                       'y',
                       df_train,
                       df_valid)
MSE
```

(d) Repeat (c) using another random seed, and report your results.
Are your results the same as what you got in (c)? Why?

```python
#New split
df_train2, df_valid2 = train_test_split(df,
                                         test_size=50,
                                         random_state=0)

#Obtain MSEs for each model
MSE2 = np.zeros(4)
for idx, degree in enumerate(range(1, 5)):
    MSE2[idx] = evalMSE([poly('x', degree)],
                       'y',
                       df_train2,
                       df_valid2)
MSE2
```

The MSEs differ between the first split (random seed = 450639) and the second split (random seed = 0) because the allocation of observations to the training set and the validation set is different. However, the general conclusion is the same: the model which produces the smallest MSEs is the quadratic model. This is the model which corresponds to the data generation process.
