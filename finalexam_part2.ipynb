## Partie 2 (/12 points)

### Vous allez générer des données simulées et les utiliser pour effectuer des prédictions à partir de régressions LASSO et OLS:


- ### (a) Génération des données et de la variable de réponse (1 pt)
1. Créez un générateur aléatoire et générez à partir d'une distribution normale standardisée :  
   - un vecteur de prédicteurs `X` de longueur \(n = 100\)  
   - un vecteur de bruit \(\epsilon\) de longueur \(n = 100\).


2. Générez ensuite la variable réponse `y` selon le modèle :  

$$y = 3 + 2 X - X^2 + 0.5 X^3 + \epsilon$$


- ### (b) Régression LASSO (6 pts)
1. Créez les prédicteurs \(X, X^2, ..., X^{10}\).
2. Entraînez un modèle **LASSO** pour prédire y à partir de ces prédicteurs.   
   - Sélectionnez automatiquement le meilleur \(\lambda\) via **10-fold cross-validation**.  
   - Utilisez la méthode **validation set** pour obtenir le MSE test.
   - Utilisez 33 % des observations pour le sous-échantillon test.
   - Utilisez la graine 436 pour chacune des deux méthodes de division de l'échantillon.
   - Incluez l'argument **max_iter=100000** dans les arguments de votre régression LASSO pour éviter d'éventuels problèmes de convergence.
3. Affichez le vecteur des coefficients entraînés.  
4. Présentez la **test-MSE**.  
5. Comparez la **root test-MSE** à l'écart-type de \(y\) pour évaluer la proportion de variabilité expliquée par le LASSO. Que constatez-vous ? 

- ### (c) Visualisation (1 pt)
1. Tracez le graphique des 100 MSE issues de la cross-validation en fonction de \(-\log(\lambda)\) en utilisant les fonctions `subplots()` et `errorbar()` comme vu en cours.  
2. Ajoutez une ligne verticale en pointillés correspondant au \(\lambda\) sélectionné.

- ### (d) Comparaison avec le modèle OLS (2 pts)
1. Entrainez un modèle OLS pour prédire y avec les mêmes prédicteurs et avec le même **validation set** (graine = 436, 33% d'observations dans l'échantillon test).
2. Présentez le vecteur de coefficients entrainés et la test-MSE.
3. Comparez la proportion de variabilité expliquée par le LASSO et à celle expliquée par le modèle OLS. Quelle est la part de variation supplémentaire expliquée par le meilleur modèle ?  
4. Comparez la norme L2 du vecteur de coefficients issus du modèle LASSO entrainé à la norme L2 du vecteur de coefficients issus du modèle OLS. A combien s'élève le ratio des deux normes ?
   > **Aide :** vous pouvez utiliser la fonction np.linalg.norm().


- ### (e) Régression LASSO sans standardisation (2 pts)
1. Ré-entraînez le modèle LASSO **sans standardiser** les prédicteurs.  
2. Comparez les coefficients et le test-MSE avec ceux obtenus précédemment.  
3. Expliquez vos observations.  
   > **Remarque :** il est normal que Python affiche des avertissements (`ConvergenceWarning`) dans cette étape.
